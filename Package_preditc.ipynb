```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("https://raw.github.com/ybifoundation/Dataset/main/Salary%20Data.csv")
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.mean()
```

```python
df.describe()
```

```python
exp = df['Experience Years']
sal = df['Salary']

plt.plot(exp, sal)
plt.show()
```

```python
plt.scatter(exp, sal)
plt.xlabel("Experience Years")
plt.ylabel("Salary")
plt.show()
```

```python
df.shape
```

```python
x = df[['Experience Years']]
y = df['Salary']

x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.8, random_state=2529)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
lr = LinearRegression()
lr.fit(x_train.values.reshape(-1, 1), y_train)
# lr = LinearRegression()
# lr.fit(x_train, y_train)
```

```python
lr.score(x_test, y_test)*100
# lr.score(x_test.values.reshape(-1, 1), y_test)
```

```python
exp = float(input("Enter your experience: "))
lr.predict([[exp]])

```

```python
e = df[['Experience Years']]
s = df[['Salary']]

plt.xlabel("Salary")
plt.ylabel('Experience Years')
plt.scatter(e, s)
plt.plot(e, lr.predict(e), color='red')
plt.show()
```
